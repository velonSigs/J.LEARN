1. jacobi and hessian Matrix




```python
import numpy as np
import matplotlib.pyplot as plt
from sympy import symbols, diff, hessian

x, y = symbols('x y')
f = x**2 * y + 3 * x + y**2

jacobi_matrix = [diff(f, x), diff(f, y)]
print("Jacobi Matrix：", jacobi_matrix)


hessian_matrix = hessian(f, (x, y))
print("Hessian Matrix：", hessian_matrix)

X, Y = np.meshgrid(np.linspace(-3, 3, 30), np.linspace(-3, 3, 30))

Z = lambda X, Y: X**2 * Y + 3 * X + Y**2

fig, ax = plt.subplots(subplot_kw={'projection': '3d'})
ax.plot_surface(X, Y, Z(X, Y), cmap='viridis')
ax.set_title('Surface plot of $f(x, y) = x^2 y + 3x + y^2$')
plt.show()
```
